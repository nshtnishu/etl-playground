```python
data = [
    (1,"John Smith","UK","2024-01-10"),
    (2,"Alice Brown","USA","2024-02-15"),
    (3,"Raj Patel","India","2024-03-01")
]

columns= ["customer_id","name","country","signup_date"]

df=spark.createDataFrame(data,columns)
df.show()
```

```python
new_data = [(4,"Emma Wilson","UK","2024-04-01"),]

new_df = spark.createDataFrame(new_data,columns)

df=df.union(new_df)

df.display()
```

```python
new_data = [(5,"Emily Johnson","UAE","2024-08-06"),(6,"Jayden Sidhu","India","2024-11-21")]

new_df = spark.createDataFrame(new_data,columns)

df=df.union(new_df)

df.display()
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
"country",
when(df.customer_id == 1,"USA").otherwise(df.country)
)
```

```python
df.display()
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
"country",
when(df.customer_id == 3,"UK")
.when(df.customer_id == 4, "UAE").otherwise(df.country)
)
```

```python
df.display()
```

```python
from pyspark.sql.functions import when

df = df.withColumn(
    "country",
    when(df.name == "John Smith", "UK")
    .when(df.name == "Raj Patel", "INDIA")
    .otherwise(df.country)
)

df.show()
```

```python
df = df.filter(df.customer_id!=1)
```
